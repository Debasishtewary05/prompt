```python
import os
from getpass import getpass
api_key=getpass("Enter your open router api key:").strip()
os.environ["OPENROUTER_API_KEY"]=api_key
print("Api key loaded sucessfully")
```

```python
from langchain_openai import ChatOpenAI

llm=ChatOpenAI(
    model="openai/gpt-oss-safeguard-20b",
    temperature=0,
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ["OPENROUTER_API_KEY"]
)
print("llm conneted sucessfully")
```

```python
response=llm.invoke("Say Hello in one sentence")
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
#prompt
prompt=ChatPromptTemplate.from_template(
    "summarize the following text in simple English:\n\n{text}")
#output parsar
parser=StrOutputParser()
#LCEL chain
chain=prompt | llm | parser
print("LCEL Chain created successfully")
```
